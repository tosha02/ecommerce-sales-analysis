```python
import pandas as pd

orders = pd.read_csv("olist_orders_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
items = pd.read_csv("olist_order_items_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
```

```python
orders_payments = pd.merge (orders, payments, on='order_id')
orders_payments.head()
```

```python
orders_payments['order_purchase_timestamp'].dtype
```

```python
orders_payments['order_purchase_timestamp'] = pd.to_datetime(
    orders_payments['order_purchase_timestamp']
)
```

```python
orders_payments['month'] = orders_payments['order_purchase_timestamp'].dt.to_period('M')
```

```python
revenue_by_month = orders_payments.groupby('month')['payment_value'].sum()
revenue_by_month
```

```python
import matplotlib.pyplot as plt

revenue_by_month.plot(figsize=(12,5))

plt.title("Revenue by Month")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
orders_items = pd.merge(orders_payments, items, on='order_id')
orders_items.head()
```

```python
orders_full = pd.merge(orders_items, products, on='product_id')
orders_full.head()
```

```python
revenue_by_category = orders_full.groupby('product_category_name')['payment_value'].sum().sort_values(ascending=False)

revenue_by_category.head(10)
```

```python
revenue_by_category.head(10).plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Product Categories by Revenue")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.show()
```

```python
orders_by_category = orders_full.groupby('product_category_name')['order_id'].count().sort_values(ascending=False)

orders_by_category.head(10)
```

```python
orders_by_category.head(10).plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Product Categories by Number of Orders")
plt.xlabel("Category")
plt.ylabel("Number of Orders")
plt.show()
```

```python
orders_per_customer = orders.groupby('customer_id')['order_id'].count()

orders_per_customer.describe()
```

```python
orders_per_customer.value_counts().head(10)
```

```python
avg_price_category = orders_full.groupby('product_category_name')['price'].mean().sort_values(ascending=False)

avg_price_category.head(10)
```

```python
avg_price_category.head(10).plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Most Expensive Product Categories")
plt.xlabel("Category")
plt.ylabel("Average Price")
plt.show()
```
